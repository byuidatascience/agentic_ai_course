## 3. Build your first agent

Run this cell. This follows [LangChain's quickstart](https://docs.langchain.com/oss/python/langchain/quickstart): a small tool the agent can call, and an agent built from a model, that tool, and a system prompt.

A response means your Python code sent a request to Gemini through LangChain, and the agent decided on its own to call the `get_weather` tool.

```python
from langchain.agents import create_agent

def get_weather(city: str) -> str:
    """Get weather for a given city."""
    return f"It's always sunny in {city}!"

agent = create_agent(
    model="google_genai:gemini-2.5-flash",
    tools=[get_weather],
    system_prompt="You are a helpful assistant",
)

result = agent.invoke(
    {"messages": [{"role": "user", "content": "What's the weather in San Francisco?"}]}
)

print(result["messages"][-1].content_blocks)
```
